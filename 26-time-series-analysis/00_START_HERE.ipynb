# Time Series Analysis - Start Here

---

Analyze and forecast sequential data, from stock prices and sensor streams to web traffic and business KPIs, using classical statistics and modern deep learning.

---

## What This Phase Is For

Time series work is different from ordinary tabular ML because order matters. The goal is to:
- Understand patterns such as trend, seasonality, and regime shifts
- Forecast future values with realistic backtesting
- Detect anomalies and operational failures in real time

## Notebooks in This Phase

| Notebook | Topic |
|----------|-------|
| `01_time_series_fundamentals.ipynb` | Stationarity, autocorrelation, decomposition |
| `02_classical_statistical_methods.ipynb` | ARIMA, SARIMA, Exponential Smoothing |
| `03_facebook_prophet.ipynb` | Prophet for business forecasting |
| `04_deep_learning_time_series.ipynb` | LSTM, TCNs, Transformers |
| `05_advanced_techniques_applications.ipynb` | Anomaly detection, multivariate forecasting |
| `06_practical_applications_exercises.ipynb` | Real datasets: stock, weather, electricity |

## Methods Overview

| Method | Strengths | When to Use |
|--------|-----------|-------------|
| **ARIMA** | Interpretable, few params | Short univariate series |
| **Prophet** | Handles holidays and seasonality well | Business KPIs |
| **LSTM** | Learns longer-range dependencies | Complex multivariate data |
| **Transformers** | Long context and parallelism | Large-scale forecasting |
| **N-BEATS / N-HiTS** | Strong forecasting accuracy | Production forecasting |

## Prerequisites
- Python and NumPy/Pandas
- Basic statistics
- Neural networks help for the deep-learning notebooks

## How To Use This Phase Well

1. Start with classical baselines before deep models.
2. Treat backtesting and evaluation as core work, not cleanup.
3. Use this phase to learn forecasting judgment, not just forecasting libraries.
