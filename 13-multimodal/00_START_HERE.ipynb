# Multimodal AI - Start Here

---

Multimodal AI extends model capability beyond text alone. This phase introduces systems that can understand or generate across text, images, audio, and related mixed-input workflows.

---

## Why This Phase Matters

A large share of practical AI products are now multimodal: image understanding, speech interfaces, document parsing, visual question answering, image generation, and multimodal retrieval. This phase helps you connect the core deep learning and LLM concepts from earlier phases to richer real-world interfaces.

## How To Use This Phase Well

- Treat each subtrack as a separate specialization rather than trying to master everything at once.
- Start with vision-language if you want the strongest connection to LLM application building.
- Use image generation and audio tracks as applied branches after the common foundations feel clear.
- Focus on use cases, limitations, and evaluation tradeoffs, not only model names.

## What You'll Learn

| Track | Materials | Focus |
|-------|-----------|-------|
| **Vision-Language** | `vision-language/` | CLIP, multimodal understanding, VLMs, multimodal RAG |
| **Image Generation** | `image-generation/` | Stable Diffusion, ControlNet, controllable generation |
| **Audio & Speech** | `audio/` | Speech recognition, text-to-speech, voice workflows |

## Prerequisites

- Neural Networks (Phase 06)
- Embeddings (Phase 05)
- RAG (Phase 08) is useful for multimodal retrieval workflows

## What Comes Next

After multimodal, continue into local LLMs, model evaluation, safety, or the specialization tracks that match the systems you want to build. This phase is best used to broaden capability after you already understand the unimodal foundations.
