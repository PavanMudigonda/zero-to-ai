## 2. Create Collection

**Embeddings:** Dense, fixed-size vector representations that capture **semantic meaning**. Unlike one-hot encoding (sparse, high-dimensional, no similarity), embeddings map similar items to nearby points in vector space.

- Words: "king" and "queen" have similar embeddings
- Sentences: Paraphrases map to nearby vectors
- Images: Similar images cluster together

Embeddings are the bridge between human-readable content and mathematical operations — enabling search, clustering, classification, and recommendation.

```python
fields = [
    FieldSchema(name="id", dtype=DataType.INT64, is_primary=True, auto_id=True),
    FieldSchema(name="text", dtype=DataType.VARCHAR, max_length=500),
    FieldSchema(name="embedding", dtype=DataType.FLOAT_VECTOR, dim=384)
]

schema = CollectionSchema(fields=fields, description="Documents")
collection = Collection(name="documents", schema=schema)

print("✅ Collection created")
```

## 3. Insert Data

Run this cell and inspect the output to verify the vector databases operations produce the expected results.

```python
entities = [
    ["Machine learning", "Deep learning", "NLP"],
    [np.random.random(384).tolist() for _ in range(3)]
]

insert_result = collection.insert(entities)
print(f"✅ Inserted {len(insert_result.primary_keys)} entities")
```

## 4. Create Index & Search

**Embeddings:** Dense, fixed-size vector representations that capture **semantic meaning**. Unlike one-hot encoding (sparse, high-dimensional, no similarity), embeddings map similar items to nearby points in vector space.

- Words: "king" and "queen" have similar embeddings
- Sentences: Paraphrases map to nearby vectors
- Images: Similar images cluster together

Embeddings are the bridge between human-readable content and mathematical operations — enabling search, clustering, classification, and recommendation.

```python
index_params = {
    "metric_type": "COSINE",
    "index_type": "IVF_FLAT",
    "params": {"nlist": 128}
}

collection.create_index(field_name="embedding", index_params=index_params)
collection.load()

search_params = {"metric_type": "COSINE", "params": {"nprobe": 10}}
query_vector = [np.random.random(384).tolist()]

results = collection.search(
    data=query_vector,
    anns_field="embedding",
    param=search_params,
    limit=3
)

print(f"✅ Search complete")
```
