## Basic usage

### Defining an expression

```python
import theano
from theano import tensor as T
x = T.vector('x')
W = T.matrix('W')
b = T.vector('b')
```

**Sigmoid Function:** Squashes any real number into the range (0, 1):

$$\sigma(x) = \frac{1}{1 + e^{-x}}$$

Properties: smooth, differentiable, S-shaped curve. At $x=0$, output is 0.5.

**Used for:** Binary classification (output = probability of positive class), gating mechanisms in LSTMs/GRUs. Downside: gradients vanish for very large/small inputs (the curve flattens), which is why ReLU often replaced it in hidden layers.

```python
dot = T.dot(x, W)
out = T.nnet.sigmoid(dot + b)
```

### Compiling a Theano function

The code below implements the next step in this deep learning workflow. See the inline comments for details on each operation.

```python
f = theano.function(inputs=[x, W], outputs=dot)
g = theano.function([x, W, b], out)
h = theano.function([x, W, b], [dot, out])
i = theano.function([x, W, b], [dot + b, out])
```

### Executing a Theano function

**Random Seed:** Setting a seed ensures **reproducibility** — the same random numbers are generated each time the code runs. This is critical in ML experiments because:
- Train/test splits will be the same
- Weight initialization will be identical
- Any randomized algorithm (dropout, data augmentation) will behave consistently

Without a fixed seed, your results would vary between runs, making it impossible to debug or compare experiments.

```python
import numpy as np
np.random.seed(42)
W_val = np.random.randn(4, 3)
x_val = np.random.rand(4)
b_val = np.ones(3)

f(x_val, W_val)
```

# Graph Transformations
## Substitution and Cloning
### The `givens` keyword

```python
x_ = T.vector('x_')
x_n = (x_ - x_.mean()) / x_.std()
f_n = theano.function([x_, W], dot, givens={x: x_n})
f_n(x_val, W_val)
```

### Cloning with replacement

The code below implements the next step in this deep learning workflow. See the inline comments for details on each operation.

```python
dot_n, out_n = theano.clone([dot, out], replace={x: (x - x.mean()) / x.std()})                        
f_n = theano.function([x, W], dot_n)                                                                  
f_n(x_val, W_val)
```

## Gradient
### Using `theano.grad`

```python
y = T.vector('y')
C = ((out - y) ** 2).sum()
dC_dW = theano.grad(C, W)
dC_db = theano.grad(C, b)
# dC_dW, dC_db = theano.grad(C, [W, b])
```

### Using the gradients

Run this cell and inspect the output to verify the deep learning operations produce the expected results.

```python
cost_and_grads = theano.function([x, W, b, y], [C, dC_dW, dC_db])
y_val = np.random.uniform(size=3)
print(cost_and_grads(x_val, W_val, b_val, y_val))
```

Run this cell and inspect the output to verify the deep learning operations produce the expected results.

```python
upd_W = W - 0.1 * dC_dW
upd_b = b - 0.1 * dC_db
cost_and_upd = theano.function([x, W, b, y], [C, upd_W, upd_b])
print(cost_and_upd(x_val, W_val, b_val, y_val))
```

The code below implements the next step in this deep learning workflow. See the inline comments for details on each operation.

```python
pydotprint(cost_and_upd, outfile='pydotprint_cost_and_upd.png')
Image('pydotprint_cost_and_upd.png', width=1000)
```

## Shared variables
### Update values

```python
C_val, dC_dW_val, dC_db_val = cost_and_grads(x_val, W_val, b_val, y_val)
W_val -= 0.1 * dC_dW_val
b_val -= 0.1 * dC_db_val

C_val, W_val, b_val = cost_and_upd(x_val, W_val, b_val, y_val)
```

### Using shared variables

**Sigmoid Function:** Squashes any real number into the range (0, 1):

$$\sigma(x) = \frac{1}{1 + e^{-x}}$$

Properties: smooth, differentiable, S-shaped curve. At $x=0$, output is 0.5.

**Used for:** Binary classification (output = probability of positive class), gating mechanisms in LSTMs/GRUs. Downside: gradients vanish for very large/small inputs (the curve flattens), which is why ReLU often replaced it in hidden layers.

```python
x = T.vector('x')
y = T.vector('y')
W = theano.shared(W_val)
b = theano.shared(b_val)
dot = T.dot(x, W)
out = T.nnet.sigmoid(dot + b)
f = theano.function([x], dot)  # W is an implicit input
g = theano.function([x], out)  # W and b are implicit inputs
print(f(x_val))
```

Evaluate `print(g(x_val))` — Jupyter renders the last expression as cell output.

```python
print(g(x_val))
```

### Updating shared variables

The code below implements the next step in this deep learning workflow. See the inline comments for details on each operation.

```python
C = ((out - y) ** 2).sum()
dC_dW, dC_db = theano.grad(C, [W, b])
upd_W = W - 0.1 * dC_dW
upd_b = b - 0.1 * dC_db

cost_and_perform_updates = theano.function(
    inputs=[x, y],
    outputs=C,
    updates=[(W, upd_W),
             (b, upd_b)])
```

The code below implements the next step in this deep learning workflow. See the inline comments for details on each operation.

```python
pydotprint(cost_and_perform_updates, outfile='pydotprint_cost_and_perform_updates.png')
Image('pydotprint_cost_and_perform_updates.png', width=1000)
```

## Extending Theano
### The easy way: Python

```python
import theano
import numpy
from theano.compile.ops import as_op

def infer_shape_numpy_dot(node, input_shapes):
    ashp, bshp = input_shapes
    return [ashp[:-1] + bshp[-1:]]

@as_op(itypes=[theano.tensor.fmatrix, theano.tensor.fmatrix],
       otypes=[theano.tensor.fmatrix], infer_shape=infer_shape_numpy_dot)
def numpy_dot(a, b):
   return numpy.dot(a, b)
```
