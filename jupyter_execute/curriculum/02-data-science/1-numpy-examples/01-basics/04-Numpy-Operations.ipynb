## Arithmetic

You can easily perform array with array arithmetic, or scalar with array arithmetic. Let's see some examples:

```python
import numpy as np
arr = np.arange(0,11)
```

Evaluate `arr` — Jupyter renders the last expression as cell output.

```python
arr
```

The code below implements the next step in this NumPy workflow. See the inline comments for details on each operation.

```python
arr + arr
```

The code below implements the next step in this NumPy workflow. See the inline comments for details on each operation.

```python
arr * arr
```

The code below implements the next step in this NumPy workflow. See the inline comments for details on each operation.

```python
arr - arr
```

can also do scalar functions - meaning function is applied to each element in the array

```python
arr + 100
```

## Universal Array Functions

Numpy comes with many [universal array functions](http://docs.scipy.org/doc/numpy/reference/ufuncs.html), which are essentially just mathematical operations you can use to perform the operation across the array. Let's show some common ones:

```python
#Taking Square Roots
np.sqrt(arr)
```

Calcualting exponential (e^)

```python
#Calcualting exponential (e^)
np.exp(arr)
```

Evaluate `np.max(arr) #same as arr.max()` — Jupyter renders the last expression as cell output.

```python
np.max(arr) #same as arr.max()
```

Evaluate `np.sin(arr)` — Jupyter renders the last expression as cell output.

```python
np.sin(arr)
```

Evaluate `np.log(arr)` — Jupyter renders the last expression as cell output.

```python
np.log(arr)
```
