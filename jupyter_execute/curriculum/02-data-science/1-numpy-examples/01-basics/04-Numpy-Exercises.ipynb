## Numpy Indexing and Selection

Now you will be given a few matrices, and be asked to replicate the resulting matrix outputs:

```python
mat = np.arange(1,26).reshape(5,5)
mat
```

Access an element or slice using indexing — `mat[2:,1:]`. Python uses zero-based indexing, so the first element is at index 0.

```python
mat[2:,1:]
```

Access an element or slice using indexing — `mat[3,4]`. Python uses zero-based indexing, so the first element is at index 0.

```python
mat[3,4]
```

SOLUTIONS includes 3rd element to return them like below:

```python
mat[:3,1:2]
```

Access an element or slice using indexing — `mat[:3,1]`. Python uses zero-based indexing, so the first element is at index 0.

```python
mat[:3,1]
```

Access an element or slice using indexing — `mat[4]`. Python uses zero-based indexing, so the first element is at index 0.

```python
mat[4]
```

Access an element or slice using indexing — `mat[3:]`. Python uses zero-based indexing, so the first element is at index 0.

```python
mat[3:]
```
