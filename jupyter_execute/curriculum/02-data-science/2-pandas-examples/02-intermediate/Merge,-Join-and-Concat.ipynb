**DataFrame Merge (Join):** Combines two DataFrames based on shared column(s), like a SQL JOIN:
- **inner**: Only matching rows from both tables
- **left/right**: All rows from one side, matching from the other
- **outer**: All rows from both, filling NaN where there's no match

Merging is essential for combining data from different sources into a single analysis-ready dataset.

```python
week1.merge(week2)
```
