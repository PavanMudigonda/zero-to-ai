The code below implements the next step in this Pandas workflow. See the inline comments for details on each operation.

```python
def rank_performance(stock_price):
    if stock_price < 120:
        return "Poor"
    elif stock_price > 120 and stock_price < 150:
        return "Average"
    else:
        return "Awesome"
    
```

**Value Counts:** Counts unique values in a column — reveals class balance, categorical distributions, and potential data quality issues.

**Apply/Map:** Apply a function to each element, row, or column. `.apply()` works on rows/columns, `.map()` on individual elements. Use for custom transformations that aren't built into Pandas.

```python
plt.style.use("ggplot")


apple["Close"].apply(rank_performance).value_counts().plot(kind="bar")
```
