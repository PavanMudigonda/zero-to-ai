**Load Data from CSV:** `pd.read_csv()` reads a comma-separated file into a Pandas DataFrame — the primary data structure for tabular data analysis. A DataFrame is like a spreadsheet: rows are observations, columns are features.

Common parameters: `header`, `index_col`, `na_values`, `dtype`, `parse_dates`. For large files, use `chunksize` or `usecols` to load only what you need.

**Handling Missing Data:** Real-world datasets almost always have missing values. Two main strategies:
- **Drop** (`.dropna()`): Remove rows/columns with missing values — simple but loses data. Appropriate when few values are missing.
- **Impute** (`.fillna()`): Replace missing values with estimated ones (mean, median, mode, or a model-based prediction). Preserves data size but introduces assumptions.

Always investigate *why* data is missing before choosing a strategy — missingness can be informative.

```python
chicago = pd.read_csv("chicago.csv", index_col="Name").dropna(how = "all")
chicago["Department"] = chicago["Department"].astype("category")
chicago.head()
```

Evaluate `chicago.index.nunique()` — Jupyter renders the last expression as cell output.

```python
chicago.index.nunique()
```

The code below implements the next step in this Pandas workflow. See the inline comments for details on each operation.

```python
chicago.index = chicago.index.str.strip().str.title()
```

The code below implements the next step in this Pandas workflow. See the inline comments for details on each operation.

```python
chicago.columns = chicago.columns.str.title()
```
