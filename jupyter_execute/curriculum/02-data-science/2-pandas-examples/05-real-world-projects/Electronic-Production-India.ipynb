**Load Data from CSV:** `pd.read_csv()` reads a comma-separated file into a Pandas DataFrame — the primary data structure for tabular data analysis. A DataFrame is like a spreadsheet: rows are observations, columns are features.

Common parameters: `header`, `index_col`, `na_values`, `dtype`, `parse_dates`. For large files, use `chunksize` or `usecols` to load only what you need.

```python
production = pd.read_csv("mobile_2.csv")
production_idx = pd.read_csv("mobile_2.csv", index_col="Item", parse_dates=True)

production_idx
```

**Transpose ($A^T$):** Flips a matrix over its diagonal — rows become columns and vice versa. If $A$ is (m×n), then $A^T$ is (n×m).

Properties: $(AB)^T = B^T A^T$, $(A^T)^T = A$. Transpose is fundamental in the Normal Equation and in computing the gradient of matrix expressions.

```python
production_t = production_idx.transpose()
production_t

path = r"mobile_t.xlsx"

writer = pd.ExcelWriter(path, engine = 'xlsxwriter')
production_t.to_excel(writer)
writer.save()
writer.close()
```
