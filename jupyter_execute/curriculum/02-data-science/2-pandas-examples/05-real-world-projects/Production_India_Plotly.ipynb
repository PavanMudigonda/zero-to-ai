**GroupBy — Split-Apply-Combine:** One of the most powerful Pandas operations. It:
1. **Splits** the data by one or more keys (categories)
2. **Applies** a function to each group independently (mean, sum, count, custom)
3. **Combines** the results back into a DataFrame

**Transpose ($A^T$):** Flips a matrix over its diagonal — rows become columns and vice versa. If $A$ is (m×n), then $A^T$ is (n×m).

Properties: $(AB)^T = B^T A^T$, $(A^T)^T = A$. Transpose is fundamental in the Normal Equation and in computing the gradient of matrix expressions.

**Preview Data:** `.head(n)` shows the first $n$ rows, `.tail(n)` the last. Always preview your data to verify it loaded correctly and understand its structure.
