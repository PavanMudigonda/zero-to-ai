# More array operations

Import the libraries needed for this section:
- **numpy**

```python
import numpy as np
```

## Expanding arrays

create a single dimension array
expand one row

The code below implements this step in the NumPy workflow.

```python
# create a single dimension array
array = np.ones((4,))
print(array)
print(array.shape)

# expand one row
expanded = array[np.newaxis, :]
expanded.shape
```

## Conditional new arrays

represent all matching conditions

The code below implements this step in the NumPy workflow.

```python
array = np.array([[3,12,11],[45,22,11],[56,15,22]])
# represent all matching conditions
print(array < 20)
```

Access an element or slice using indexing — `array[array < 20]`. Python uses zero-based indexing, so the first element is at index 0.

```python
# produce all matching values
array[array < 20]
```

## Slicing

Similar to Python lists, but with more features and capabilities

```python
a = np.arange(6)
a
```

Access an element or slice using indexing — `a[1:4]`. Python uses zero-based indexing, so the first element is at index 0.

```python
# give a range
a[1:4]
```

## Stacking

horizontally
must have same number of dimensions!

The code below implements this step in the NumPy workflow.

```python
# horizontally
# must have same number of dimensions!
first = np.full((2, 12), 2)
print(first)
second = np.full((2, 7), 2)
print(second)
np.hstack((first, second))
```

stacking vertically

The code below implements this step in the NumPy workflow.

```python
# stacking vertically
first = np.full((3,3), 1)
print(first)
second = np.full((3,3), 2)
print(second)
np.vstack((first, second))
```


## Splitting arrays

The code below implements the next step in this NumPy workflow. See the inline comments for details on each operation.

```python
array = np.arange(1, 11)
array
```

- split into two equal parts
- warning it needs to allow for an equal division!

```python
# split into two equal parts
# warning it needs to allow for an equal division!
np.hsplit(array, 2)
```

this allows for "unpacking"

The code below implements this step in the NumPy workflow.

```python
# this allows for "unpacking"
first, second = np.hsplit(array, 2)
print(first)
print(second)
```
