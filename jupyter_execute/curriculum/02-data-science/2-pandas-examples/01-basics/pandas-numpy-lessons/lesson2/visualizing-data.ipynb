# Visualizing data
`matplotlib` is able to produce meaningful data plots right in a notebook with little effort using Pandas DataFrames. This notebook covers a few of those capabilities.

```python
# Load your dataframe
import pandas as pd
csv_url = "https://raw.githubusercontent.com/paiml/wine-ratings/main/wine-ratings.csv"
df = pd.read_csv(csv_url, index_col=0)
df.drop('grape', axis=1, inplace=True)
df = df.dropna()
df.head()
```

Import `matplotlib` and set the `inline` magic line

```python
!pip install matplotlib
```

**Apply/Map:** Apply a function to each element, row, or column. `.apply()` works on rows/columns, `.map()` on individual elements. Use for custom transformations that aren't built into Pandas.

```python
df['rating'] = df['rating'].apply(lambda x: int(x))
```

Evaluate `df.plot()` — Jupyter renders the last expression as cell output.

```python
df.plot()
```

Access an element or slice using indexing — `df["variety"].plot()`. Python uses zero-based indexing, so the first element is at index 0.

```python
df["variety"].plot()
```

The code below implements the next step in this NumPy workflow. See the inline comments for details on each operation.

```python
df.plot.scatter(x="rating", y="variety", alpha=0.5)
```
