## The .duplicated() Method

**Load Data from CSV:** `pd.read_csv()` reads a comma-separated file into a Pandas DataFrame — the primary data structure for tabular data analysis. A DataFrame is like a spreadsheet: rows are observations, columns are features.

Common parameters: `header`, `index_col`, `na_values`, `dtype`, `parse_dates`. For large files, use `chunksize` or `usecols` to load only what you need.

```python
df = pd.read_csv("employees.csv", parse_dates = ["Start Date", "Last Login Time"])
df["Senior Management"] = df["Senior Management"].astype("bool")
df["Gender"] = df["Gender"].astype("category")
df.sort_values("First Name", inplace=True)
df.head(3)
```

The code below implements the next step in this Pandas workflow. See the inline comments for details on each operation.

```python
mask = ~df["First Name"].duplicated(keep=False)
df[mask]
```
