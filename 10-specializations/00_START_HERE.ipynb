# AI Specializations - Start Here

---

This phase is a specialization hub, not one single linear notebook track. Choose one path based on the kind of work you want to get good at, then open that subfolder's `00_START_HERE.ipynb`.

---

## Choose One Track First

| Track | Directory | Best For |
|-------|-----------|----------|
| **Computer Vision** | `computer-vision/` | Image understanding, detection, segmentation, visual retrieval |
| **NLP** | `nlp/` | Extraction, summarization, translation, document workflows |
| **AI Agents** | `ai-agents/` | Tool use, workflows, planning, multi-agent systems |

## Actual Learning Entry Points

Open one of these first:

- `computer-vision/00_START_HERE.ipynb`
- `nlp/00_START_HERE.ipynb`
- `ai-agents/00_START_HERE.ipynb`

## How To Choose

- **Building product workflows or copilots?** Start with **AI Agents**.
- **Working with images, OCR, or multimodal search?** Start with **Computer Vision**.
- **Working with documents, language tasks, or text pipelines?** Start with **NLP**.

## Prerequisites

- Embeddings and retrieval basics from earlier phases
- Neural network basics for the Computer Vision track
- Prompting and RAG familiarity for the AI Agents track
- Solid Python fundamentals for all tracks

## How To Use This Phase Well

1. Pick one track first instead of sampling all three.
2. Complete one meaningful project inside that track.
3. Use the parent specialization README for broader context, but treat the subtrack README and `00_START_HERE.ipynb` as the real entry point.
