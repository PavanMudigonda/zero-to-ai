# Causal Inference - Start Here

---

Causal inference is about estimating what actually changes outcomes, not just what correlates with them. This phase is especially useful for experimentation, policy decisions, product analysis, and high-stakes decision systems.

---

## Why This Phase Matters

Standard machine learning is usually optimized for prediction. Causal inference is about intervention: what happens if you change something. That difference matters for ads, pricing, medicine, policy, product decisions, and any setting where action matters more than forecasting alone.

## How To Use This Phase Well

- Treat this as a statistics-heavy specialization, not a mandatory prerequisite for the LLM path.
- Spend time on assumptions, not just estimators.
- Keep experiments, DAGs, and observational methods conceptually separate.
- Use this phase if your goals include experimentation, decision science, product analytics, or policy-facing ML.
