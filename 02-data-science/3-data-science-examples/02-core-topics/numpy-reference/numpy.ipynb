## Reshape and Update In-Place

Run this cell and inspect the output to verify the NumPy operations produce the expected results.

```python
e = np.arange(12)
print(e)
```

f is a view of contents of e

The code below implements this step in the NumPy workflow.

```python
# f is a view of contents of e
f = e.reshape(3, 4)
print(f)
```

### Set values of e from index 5 onwards to 0

The code below implements this step in the NumPy array operations workflow.

```python
# Set values of e from index 5 onwards to 0
e[5:] = 0
print(e)
```

f is also updated

```python
# f is also updated
f
```

OWNDATA shows f does not own its data

```python
# OWNDATA shows f does not own its data
f.flags
```

## Create Sample Data

Import the libraries needed for this section:
- **pylab**
- **seaborn**

```python
%matplotlib inline

import pylab as plt
import seaborn

seaborn.set()
```

**Line Plot:** Connects data points with lines — ideal for showing trends over a continuous variable (often time). In ML, line plots are commonly used for:
- Training/validation loss curves (to detect overfitting)
- Learning rate schedules
- Time series data

```python
# Create evenly spaced numbers over the specified interval
x = np.linspace(0, 2, 10)
plt.plot(x, 'o-');
plt.show()
```

**Scatter Plot:** Displays individual data points as dots on a 2D plane. Each point's position is determined by its x and y values. Color and size can encode additional dimensions.

**Use for:** Exploring relationships between two continuous variables, spotting clusters, identifying outliers, and visualizing model predictions vs. actual values.

```python
# Create sample data, add some noise
x = np.random.uniform(1, 100, 1000)
y = np.log(x) + np.random.normal(0, .3, 1000)

plt.scatter(x, y)
plt.show()
```
