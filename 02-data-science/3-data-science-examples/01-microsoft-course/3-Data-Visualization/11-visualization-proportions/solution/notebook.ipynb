# 🍄 Mushroom Proportions

Import the mushroom dataset

**Load Data from CSV:** `pd.read_csv()` reads a comma-separated file into a Pandas DataFrame — the primary data structure for tabular data analysis. A DataFrame is like a spreadsheet: rows are observations, columns are features.

Common parameters: `header`, `index_col`, `na_values`, `dtype`, `parse_dates`. For large files, use `chunksize` or `usecols` to load only what you need.

```python
import pandas as pd
import matplotlib.pyplot as plt
mushrooms = pd.read_csv('../../../data/mushrooms.csv')
mushrooms.head()
```

# Pie chart

Create a pie chart displaying the proportion of Poisonous vs. Edible mushrooms

```python
print(mushrooms.select_dtypes(["object"]).columns)
```

The code below implements the next step in this data science workflow. See the inline comments for details on each operation.

```python
cols = mushrooms.select_dtypes(["object"]).columns
mushrooms[cols] = mushrooms[cols].astype('category')
```

**GroupBy — Split-Apply-Combine:** One of the most powerful Pandas operations. It:
1. **Splits** the data by one or more keys (categories)
2. **Applies** a function to each group independently (mean, sum, count, custom)
3. **Combines** the results back into a DataFrame

Example: `df.groupby('city')['price'].mean()` computes the average price per city.

```python
edibleclass=mushrooms.groupby(['class']).count()
edibleclass
```

The code below implements the next step in this data science workflow. See the inline comments for details on each operation.

```python
labels=['Edible','Poisonous']
plt.pie(edibleclass['population'],labels=labels,autopct='%.1f %%')
plt.title('Edible?')
plt.show()
```

# Donut chart

**GroupBy — Split-Apply-Combine:** One of the most powerful Pandas operations. It:
1. **Splits** the data by one or more keys (categories)
2. **Applies** a function to each group independently (mean, sum, count, custom)
3. **Combines** the results back into a DataFrame

Example: `df.groupby('city')['price'].mean()` computes the average price per city.

```python
habitat=mushrooms.groupby(['habitat']).count()
habitat
```

Adding Title of chart

The code below implements this step in the data science workflow.

```python
  
labels=['Grasses','Leaves','Meadows','Paths','Urban','Waste','Wood']

plt.pie(habitat['class'], labels=labels,
        autopct='%1.1f%%', pctdistance=0.85)
  
center_circle = plt.Circle((0, 0), 0.40, fc='white')
fig = plt.gcf()

fig.gca().add_artist(center_circle)
  
# Adding Title of chart
plt.title('Mushroom Habitats')
  
plt.show()
```

**Create a DataFrame:** Pandas DataFrames are 2D labeled data structures — they're the backbone of data science in Python. Each column can hold a different data type (numbers, strings, dates). DataFrames support powerful operations: filtering, grouping, merging, reshaping, and aggregation.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pywaffle import Waffle
  
# creation of a dataframe


data ={'color': ['brown', 'buff', 'cinnamon', 'green', 'pink', 'purple', 'red', 'white', 'yellow'],
    'amount': capcolor['class']
     }
  
df = pd.DataFrame(data)
  
# To plot the waffle Chart
fig = plt.figure(
    FigureClass = Waffle,
    rows = 100,
    values = df.amount,
    labels = list(df.color),
    figsize = (30,30),
    colors=["brown", "tan", "maroon", "green", "pink", "purple", "red", "whitesmoke", "yellow"],
)

```
