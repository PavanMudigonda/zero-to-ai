```python
import pandas as pd
import glob

#Loading the dataset
path = '../../data/emails.csv'
email_df = pd.read_csv(path)
```

**DataFrame.describe():** Quick statistical summary — count, mean, std, min, max, and quartiles for each numeric column. First thing to run on new data to understand ranges, detect outliers, and check for missing values.

```python
# Using Describe on the email dataset
print(email_df.describe())
```

Sampling 10 emails

```python
# Sampling 10 emails
print(email_df.sample(10))
```

Returns rows where there are more occurrences of "to" than "the"

```python
# Returns rows where there are more occurrences of "to" than "the"
print(email_df.query('the < to'))
```
