##### Linear Algebra

**Matrix Multiplication:** The core operation in nearly all ML. When you multiply matrices $A$ (m×n) and $B$ (n×p), each element of the result is a **dot product** — a row of $A$ dotted with a column of $B$:

$$C_{ij} = \sum_{k=1}^{n} A_{ik} \cdot B_{kj}$$

**Why it matters:**
- Every neural network layer computes $\text{output} = W \cdot \text{input} + b$
- Transformers use matrix multiplication extensively for attention scores
- The `@` operator in Python/NumPy is shorthand for `np.matmul()`

```python
a = np.ones((2,3))
print(a)

b = np.full((3,2), 2)
print(b)

np.matmul(a,b)
```

**Determinant:** A scalar that measures how a matrix scales area (in 2D) or volume (in higher dimensions). For a 2×2 matrix $\begin{bmatrix}a & b\\c & d\end{bmatrix}$, the determinant is $ad - bc$.

- **det = 0**: The matrix is singular — it collapses space into a lower dimension (no inverse exists)
- **det > 0**: Preserves orientation
- **det < 0**: Flips orientation (like a mirror)

In ML, determinants appear in multivariate Gaussian distributions and in checking whether a system of equations has a unique solution.

```python
# Find the determinant
c = np.identity(3)
np.linalg.det(c)
```
