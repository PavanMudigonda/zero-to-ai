## Indexing a 2D array (matrices)

The general format is **arr_2d[row][col]** or **arr_2d[row,col]**. I recommend usually using the comma notation for clarity.

```python
arr_2d = np.array(([5,10,15],[20,25,30],[35,40,45]))

#Show
arr_2d
```

2 ways to grab values from matrix - first is double bracket notation, similar to grabbing from python dict

```python
#Indexing row
arr_2d[1]
```

Access an element or slice using indexing — `arr_2d[1][0]`. Python uses zero-based indexing, so the first element is at index 0.

```python
# Format is arr_2d[row][col] or arr_2d[row,col]

# Getting individual element value
arr_2d[1][0]
```

RECOMMENDED: comma notation

Access an element or slice using indexing — `arr_2d[1,0]`. Python uses zero-based indexing, so the first element is at index 0.

```python
# Getting individual element value
arr_2d[1,0]
```

grabbing section of a matrix

Access an element or slice using indexing — `arr_2d[:2,1:]`. Python uses zero-based indexing, so the first element is at index 0.

```python
# 2D array slicing

#Shape (2,2) from top right corner
arr_2d[:2,1:]
```

Access an element or slice using indexing — `arr_2d[2]`. Python uses zero-based indexing, so the first element is at index 0.

```python
#Shape bottom row
arr_2d[2]
```

Access an element or slice using indexing — `arr_2d[2,:]`. Python uses zero-based indexing, so the first element is at index 0.

```python
#Shape bottom row
arr_2d[2,:]
```

### Fancy Indexing

Fancy indexing allows you to select entire rows or columns out of order,to show this, let's quickly build out a numpy array:

```python
#Set up matrix
arr2d = np.zeros((10,10))
```

Length of array

The code below implements this step in the NumPy workflow.

```python
#Length of array
arr_length = arr2d.shape[1]
```

Set up array

The code below implements this step in the NumPy workflow.

```python
#Set up array

for i in range(arr_length):
    arr2d[i] = i
    
arr2d
```

Fancy indexing allows the following

Access an element or slice using indexing — `arr2d[[2,4,6,8]]`. Python uses zero-based indexing, so the first element is at index 0.

```python
arr2d[[2,4,6,8]]
```

Access an element or slice using indexing — `arr2d[[6,4,2,7]]`. Python uses zero-based indexing, so the first element is at index 0.

```python
#Allows in any order
arr2d[[6,4,2,7]]
```
