```python
import plotly
from plotly.graph_objs import Scatter, Layout
import plotly.graph_objs as go


stepdata = [go.Bar(x=walking["date"].dt.date, y=walking["metres"])]


plotly.offline.plot({ 'data': stepdata,
            'layout': {
               'title': 'Walking (metres)',
               'xaxis': {
                 'title': 'Creation Date'},
               'yaxis': {
                'title': 'Value '}
        }})
```

The code below implements the next step in this Pandas workflow. See the inline comments for details on each operation.

```python
def km_to_m(km):
    return int(km*1000)
```

**Apply/Map:** Apply a function to each element, row, or column. `.apply()` works on rows/columns, `.map()` on individual elements. Use for custom transformations that aren't built into Pandas.

```python
walking["metres"] = walking["value"].apply(km_to_m)
```

Access an element or slice using indexing — `walking["metres"].mean()`. Python uses zero-based indexing, so the first element is at index 0.

```python
walking["metres"].mean()
```
