## Reshaping arrays

The code below implements the next step in this NumPy workflow. See the inline comments for details on each operation.

```python
array = np.ones((3,4))
array
```

Evaluate `array.reshape((4,3))` — Jupyter renders the last expression as cell output.

```python
array.reshape((4,3))
```

```python
# reshaping has to keep the same number of items
array.reshape((12,1))
```
