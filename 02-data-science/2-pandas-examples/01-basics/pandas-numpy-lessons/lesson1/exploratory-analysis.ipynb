## Exploratory Analysis

One of the main tasks to perform with Pandas is exploratory analysis. Looking at data, finding what is useful or potentially wrong with it so that you can clean it up are core practices of a data scientist and data engineer.

## Create a Pandas Dataframe 
Load a CSV to start working with the data and performing exploratory analysis.

```python
import pandas as pd
csv_url = "https://raw.githubusercontent.com/paiml/wine-ratings/main/wine-ratings.csv"
df = pd.read_csv(csv_url, index_col=0)
```

**Preview Data:** `.head(n)` shows the first $n$ rows, `.tail(n)` the last. Always preview your data to verify it loaded correctly and understand its structure.

```python
# The most common operation is with .head() 
df.head(15)
```

**DataFrame.describe():** Quick statistical summary — count, mean, std, min, max, and quartiles for each numeric column. First thing to run on new data to understand ranges, detect outliers, and check for missing values.

```python

# Now lets get a description of the data
df.describe()
```

**DataFrame.info():** Shows column names, data types, non-null counts, and memory usage. Quickly reveals: which columns have missing values, which are the wrong type (e.g., numbers stored as strings), and how much memory the data uses.

```python
# You can also get metadata about the dataset with .info()
df.info()
```

sort based on some condition

The code below implements this step in the NumPy workflow.

```python
# sort based on some condition
df.sort_values(by="rating", ascending=False).head()
```

Remove any newlines or carriage returns

**Preview Data:** `.head(n)` shows the first $n$ rows, `.tail(n)` the last. Always preview your data to verify it loaded correctly and understand its structure.

```python
df = df.replace({"\r": ""}, regex=True)
df = df.replace({"\n": " "}, regex=True)
df.head(10)
```

the grape is not a very good column, lets remove it and describe it again

The code below implements this step in the NumPy workflow.

```python
# the grape is not a very good column, lets remove it and describe it again
df.drop(['grape'], axis=1, inplace=True)
df.describe()
```

**GroupBy — Split-Apply-Combine:** One of the most powerful Pandas operations. It:
1. **Splits** the data by one or more keys (categories)
2. **Applies** a function to each group independently (mean, sum, count, custom)
3. **Combines** the results back into a DataFrame

Example: `df.groupby('city')['price'].mean()` computes the average price per city.

```python
# Specific operations by method. Like .mean()
df.groupby("region").mean(numeric_only=True)
```
