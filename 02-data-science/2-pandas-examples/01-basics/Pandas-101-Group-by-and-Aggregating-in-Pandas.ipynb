**Load Data from CSV:** `pd.read_csv()` reads a comma-separated file into a Pandas DataFrame — the primary data structure for tabular data analysis. A DataFrame is like a spreadsheet: rows are observations, columns are features.

Common parameters: `header`, `index_col`, `na_values`, `dtype`, `parse_dates`. For large files, use `chunksize` or `usecols` to load only what you need.

**GroupBy — Split-Apply-Combine:** One of the most powerful Pandas operations. It:
1. **Splits** the data by one or more keys (categories)
2. **Applies** a function to each group independently (mean, sum, count, custom)
3. **Combines** the results back into a DataFrame

Example: `df.groupby('city')['price'].mean()` computes the average price per city.

```python
group_by_frame = df.groupby('Base Flavor')
```

Evaluate `group_by_frame.mean()` — Jupyter renders the last expression as cell output.

```python
group_by_frame.mean()
```
