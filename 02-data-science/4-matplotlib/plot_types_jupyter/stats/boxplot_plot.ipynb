# Box and Whisker Plots

A box plot (also called a box-and-whisker plot) summarizes the distribution of a dataset through its quartiles: the box spans from the 25th to the 75th percentile (the interquartile range, or IQR), the line inside the box marks the median, and the whiskers extend to show the range of the data (typically 1.5x IQR). The `Axes.boxplot()` method accepts a 2D array where each column becomes one box.

**Why this matters for data science:** Box plots are one of the most important statistical visualizations in data science. They provide a compact, standardized summary of distribution shape, central tendency, spread, and outliers -- all in a single glyph. They are used extensively for comparing feature distributions across groups, detecting outliers, checking normality assumptions before modeling, and summarizing cross-validation results. The example customizes the appearance heavily using `patch_artist=True` (to enable filled boxes), property dictionaries like `boxprops` and `medianprops` for colors and line widths, and `showfliers=False` to hide outlier points. The data is generated with `np.random.normal()` using different means (3, 5, 4) and standard deviations (1.25, 1.00, 1.25) for each of the three distributions, creating a realistic comparison scenario.

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('_mpl-gallery')

# make data:
np.random.seed(10)
D = np.random.normal((3, 5, 4), (1.25, 1.00, 1.25), (100, 3))

# plot
fig, ax = plt.subplots()
VP = ax.boxplot(D, positions=[2, 4, 6], widths=1.5, patch_artist=True,
                showmeans=False, showfliers=False,
                medianprops={"color": "white", "linewidth": 0.5},
                boxprops={"facecolor": "C0", "edgecolor": "white",
                          "linewidth": 0.5},
                whiskerprops={"color": "C0", "linewidth": 1.5},
                capprops={"color": "C0", "linewidth": 1.5})

ax.set(xlim=(0, 8), xticks=np.arange(1, 8),
       ylim=(0, 8), yticks=np.arange(1, 8))

plt.show()
```
