# Error Bar Plots

Error bar plots display data points with bars extending above and below (or left and right) to indicate uncertainty, variability, or confidence intervals. The `Axes.errorbar()` method takes x and y arrays for the central values, plus `yerr` (and optionally `xerr`) arrays specifying the error magnitude in each direction.

**Why this matters for data science:** Reporting results without uncertainty estimates is one of the most common mistakes in data analysis. Error bars communicate the reliability of measurements and are essential for experimental results, model evaluation metrics (accuracy with confidence intervals), and any comparison where you need to know whether differences are meaningful. The `fmt='o'` parameter draws markers only (no connecting line), `capsize=6` adds horizontal caps at the ends of the error bars for readability, and `linewidth=2` makes the bars visually prominent. The yerr values here (0.9, 1.2, 0.5) are asymmetric across points, reflecting the common real-world situation where different measurements have different levels of uncertainty. You can also pass a 2x3 array to yerr for asymmetric error bars (different upper and lower bounds).

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('_mpl-gallery')

# make data:
np.random.seed(1)
x = [2, 4, 6]
y = [3.6, 5, 4.2]
yerr = [0.9, 1.2, 0.5]

# plot:
fig, ax = plt.subplots()

ax.errorbar(x, y, yerr, fmt='o', linewidth=2, capsize=6)

ax.set(xlim=(0, 8), xticks=np.arange(1, 8),
       ylim=(0, 8), yticks=np.arange(1, 8))

plt.show()
```
