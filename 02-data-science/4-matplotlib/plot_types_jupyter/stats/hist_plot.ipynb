### Histograms with `hist(x)`

A histogram groups continuous data into bins and displays the count (or frequency) of observations in each bin as a bar. The `Axes.hist()` method both computes the binning and renders the visualization in a single call, making it the quickest way to inspect the shape of a distribution.

**Why this matters for data science:** Histograms are typically the first plot you create when exploring a new dataset -- they immediately reveal the distribution shape (normal, skewed, bimodal, uniform), the data range, and potential outliers. They are essential during exploratory data analysis (EDA), feature engineering (checking if a log transform would help), and model diagnostics (examining residual distributions). The `bins=8` parameter controls the number of bins, which is a critical choice: too few bins obscure the distribution shape, while too many create noisy, jagged plots. Rules of thumb like Sturges' rule or the Freedman-Diaconis rule can help, but visual inspection is often best. The `edgecolor="white"` with small `linewidth` creates clean separation between bars, improving readability. The data here is drawn from a normal distribution centered at 4, and the resulting histogram should approximate the characteristic bell curve shape.

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('_mpl-gallery')

# make data
np.random.seed(1)
x = 4 + np.random.normal(0, 1.5, 200)

# plot:
fig, ax = plt.subplots()

ax.hist(x, bins=8, linewidth=0.5, edgecolor="white")

ax.set(xlim=(0, 8), xticks=np.arange(1, 8),
       ylim=(0, 56), yticks=np.linspace(0, 56, 9))

plt.show()
```
