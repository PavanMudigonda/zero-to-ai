# 3D Scatter Plots

A 3D scatter plot places individual data points in three-dimensional space, making it a direct extension of the 2D scatter plot with an added z-axis. Each point is defined by three coordinate values, and you can optionally map additional variables to marker size, color, or shape.

**Why this matters for data science:** 3D scatter plots are one of the most frequently used 3D visualizations in ML and data analysis. They are invaluable for exploring relationships among three continuous features simultaneously -- for instance, visualizing clusters identified by K-means or DBSCAN, inspecting the output of dimensionality reduction techniques like PCA or t-SNE (when reduced to 3 components), or examining the joint distribution of three features in exploratory data analysis. The example uses `np.random.default_rng()` with a fixed seed for reproducibility, generating uniform random data with `rng.uniform()` across different ranges for each axis. Reproducible random state is a best practice in data science workflows to ensure consistent results across runs.

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('_mpl-gallery')

# Make data
np.random.seed(19680801)
n = 100
rng = np.random.default_rng()
xs = rng.uniform(23, 32, n)
ys = rng.uniform(0, 100, n)
zs = rng.uniform(-50, -25, n)

# Plot
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.scatter(xs, ys, zs)

ax.set(xticklabels=[],
       yticklabels=[],
       zticklabels=[])

plt.show()
```
