# 3D Line Plots

The 3D `plot()` method draws a continuous line through a sequence of points in three-dimensional space, extending matplotlib's standard 2D line plot with a z-axis. You pass three arrays of equal length representing the x, y, and z coordinates of each point, and the method connects them in order to form a 3D curve.

**Why this matters for data science:** 3D line plots are essential for visualizing trajectories and parametric curves -- think of tracking an object's position over time in three dimensions, plotting the path of gradient descent through a loss landscape, or displaying phase-space trajectories in dynamical systems. The example generates a helix by computing sine and cosine as functions of a linearly spaced parameter, producing the classic spiral shape. The key setup step is `subplot_kw={"projection": "3d"}`, which tells matplotlib to create a 3D-capable axes object from `mpl_toolkits.mplot3d` instead of the default 2D axes.

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('_mpl-gallery')

# Make data
n = 100
xs = np.linspace(0, 1, n)
ys = np.sin(xs * 6 * np.pi)
zs = np.cos(xs * 6 * np.pi)

# Plot
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.plot(xs, ys, zs)

ax.set(xticklabels=[],
       yticklabels=[],
       zticklabels=[])

plt.show()
```
