# Triangulated Surface Plots

A triangulated surface plot constructs a 3D surface from irregularly spaced data points by automatically computing a Delaunay triangulation. Unlike `plot_surface()` which requires data on a regular grid, `plot_trisurf()` accepts 1D arrays of x, y, and z coordinates and connects them into triangular facets to approximate a continuous surface.

**Why this matters for data science:** Real-world data is rarely sampled on a perfect grid. Sensor networks, geographic surveys, and experimental measurements typically produce irregularly spaced observations. Triangulated surfaces let you visualize these datasets without the preprocessing step of interpolating onto a regular grid. The example generates data in polar coordinates (radii and angles), converts to Cartesian coordinates using `np.cos()` and `np.sin()`, and computes z as `sin(-x*y)`. The `[..., np.newaxis]` broadcasting trick efficiently creates the full grid of (radius, angle) combinations. This polar-to-Cartesian conversion pattern is common when working with circular or radial data in scientific computing.

```python
import matplotlib.pyplot as plt
import numpy as np

from matplotlib import cm

plt.style.use('_mpl-gallery')

n_radii = 8
n_angles = 36

# Make radii and angles spaces
radii = np.linspace(0.125, 1.0, n_radii)
angles = np.linspace(0, 2*np.pi, n_angles, endpoint=False)[..., np.newaxis]

# Convert polar (radii, angles) coords to cartesian (x, y) coords.
x = np.append(0, (radii*np.cos(angles)).flatten())
y = np.append(0, (radii*np.sin(angles)).flatten())
z = np.sin(-x*y)

# Plot
fig, ax = plt.subplots(subplot_kw={'projection': '3d'})
ax.plot_trisurf(x, y, z, vmin=z.min() * 2, cmap=cm.Blues)

ax.set(xticklabels=[],
       yticklabels=[],
       zticklabels=[])

plt.show()
```
