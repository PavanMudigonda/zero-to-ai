# Scatter Plots

A scatter plot displays individual data points as markers in 2D space, with optional encoding of additional variables through marker size and color. The `Axes.scatter()` method takes x and y arrays, plus optional `s` (size) and `c` (color) arrays to map up to four dimensions of data onto a single 2D plot.

**Why this matters for data science:** Scatter plots are the primary tool for exploring relationships between two continuous variables -- correlation analysis, regression diagnostics, cluster visualization, and anomaly detection all rely on them. The ability to encode size and color as additional dimensions (`s=sizes, c=colors`) is powerful: for example, you might plot predicted vs. actual values (x, y), with point size representing sample weight and color representing prediction error. The `vmin` and `vmax` parameters control the range of the color mapping, ensuring consistent color interpretation across plots. The example uses `np.random.normal()` to generate normally distributed x and y values centered at 4, simulating the kind of clustered data you would encounter in real datasets.

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('_mpl-gallery')

# make the data
np.random.seed(3)
x = 4 + np.random.normal(0, 2, 24)
y = 4 + np.random.normal(0, 2, len(x))
# size and color:
sizes = np.random.uniform(15, 80, len(x))
colors = np.random.uniform(15, 80, len(x))

# plot
fig, ax = plt.subplots()

ax.scatter(x, y, s=sizes, c=colors, vmin=0, vmax=100)

ax.set(xlim=(0, 8), xticks=np.arange(1, 8),
       ylim=(0, 8), yticks=np.arange(1, 8))

plt.show()
```
