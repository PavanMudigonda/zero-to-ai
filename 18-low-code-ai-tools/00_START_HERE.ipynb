# Low-Code AI Tools — Start Here

---

Build and deploy AI applications quickly with Gradio demos, Streamlit dashboards, Hugging Face Spaces, and AutoML platforms.

This phase is best used as a fast application-building module. It is practical and useful, but it is not meant to replace deeper engineering phases like MLOps, agents, or evaluation.

---

## Why Low-Code Matters

Getting AI into users' hands fast is often more valuable than perfect engineering. These tools let you prototype quickly, share publicly, and validate whether an idea is worth deeper investment.

## Notebooks in This Phase

| Notebook | Topic |
|----------|-------|
| `01_gradio_basics.ipynb` | Gradio — instant AI UIs with one function |
| `02_streamlit_apps.ipynb` | Streamlit — data apps with pure Python |
| `03_huggingface_spaces.ipynb` | Deploy to HF Spaces |
| `04_automl_platforms.ipynb` | AutoML: H2O, AutoGluon, FLAML |
| `05_end_to_end_project.ipynb` | Build and deploy a complete AI app |

## Tool Comparison

| Tool | Best For | Deploy Where |
|------|----------|-------------|
| **Gradio** | ML demos, model showcases | HF Spaces, local |
| **Streamlit** | Data dashboards, internal tools | Streamlit Cloud, local |
| **HF Spaces** | Public model demos | huggingface.co/spaces |
| **AutoML** | Non-experts building classifiers | Local, cloud |

## Prerequisites
- Python basics
- Any LLM or ML model to wrap

## How To Use This Phase Well

1. Use this phase to get an application in front of users quickly.
2. Pick one interface path first instead of trying every tool equally.
3. Treat the capstone as a prototype you can later harden with MLOps or agent workflows.
