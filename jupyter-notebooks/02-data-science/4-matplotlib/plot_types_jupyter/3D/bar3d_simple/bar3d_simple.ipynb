# 3D Bar Charts

A 3D bar chart extends the familiar 2D bar chart into three dimensions, allowing you to represent data across two categorical axes with bar heights encoding a third numeric variable. The `Axes3D.bar3d()` method requires six parameters: **x, y, z** define the anchor point of each bar (where z is typically 0 for bars rising from the floor), while **dx, dy, dz** control the width, depth, and height of each bar respectively.

**Why this matters for data science:** 3D bar charts are useful when comparing values across two independent categorical variables simultaneously -- for example, sales figures across different products and regions. In ML contexts, you might use them to visualize confusion matrices or hyperparameter grid search results. The `np.ones_like()` call here creates uniform bar widths, but you can vary these to encode additional information. Note that while 3D plots look impressive, they can sometimes obscure data through occlusion, so use them judiciously when a heatmap or grouped bar chart would not suffice.

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('_mpl-gallery')

# Make data
x = [1, 1, 2, 2]
y = [1, 2, 1, 2]
z = [0, 0, 0, 0]
dx = np.ones_like(x)*0.5
dy = np.ones_like(x)*0.5
dz = [2, 3, 1, 4]

# Plot
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.bar3d(x, y, z, dx, dy, dz)

ax.set(xticklabels=[],
       yticklabels=[],
       zticklabels=[])

plt.show()
```
