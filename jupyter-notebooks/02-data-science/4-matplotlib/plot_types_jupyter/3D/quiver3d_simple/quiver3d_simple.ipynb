# 3D Quiver (Vector Field) Plots

A 3D quiver plot displays arrows (vectors) at points in three-dimensional space, where each arrow's direction and length represent a vector quantity. The method takes six array arguments: **X, Y, Z** specify the origin points of each arrow, while **U, V, W** define the vector components (direction and magnitude) at each point.

**Why this matters for data science:** Quiver plots visualize vector fields, which appear throughout science and engineering -- electromagnetic fields, fluid flow velocity, force distributions, and gradient fields in optimization. In machine learning, you can use 3D quiver plots to visualize gradient vectors across a parameter space, helping you understand how optimization algorithms navigate loss surfaces. The example uses `np.meshgrid()` to create a regular 3D grid of points, then computes vector components as simple functions of position. Setting W to zero creates a purely horizontal flow field, which is a useful simplification for understanding the X-Y dynamics before adding vertical complexity.

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('_mpl-gallery')

# Make data
n = 4
x = np.linspace(-1, 1, n)
y = np.linspace(-1, 1, n)
z = np.linspace(-1, 1, n)
X, Y, Z = np.meshgrid(x, y, z)
U = (X + Y)/5
V = (Y - X)/5
W = Z*0


# Plot
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.quiver(X, Y, Z, U, V, W)

ax.set(xticklabels=[],
       yticklabels=[],
       zticklabels=[])

plt.show()
```
