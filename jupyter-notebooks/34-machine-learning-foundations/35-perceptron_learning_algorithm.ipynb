## Core IdeaFor a misclassified sample $(x_i,y_i)$ with $y_i\in\{-1,1\}$, update $w\leftarrow w+\eta y_i x_i$ and $b\leftarrow b+\eta y_i$.

```python
import numpy as np
X = np.array([[0,0], [0,1], [1,0], [1,1]], dtype=float)
y = np.array([-1, -1, -1, 1])  # AND
w = np.zeros(2); b = 0.0
for epoch in range(20):
    errors = 0
    for features, target in zip(X, y):
        prediction = 1 if features @ w + b >= 0 else -1
        if prediction != target:
            w += target * features; b += target; errors += 1
    if errors == 0: break
print('weights:', w, 'bias:', b, 'predictions:', np.where(X @ w + b >= 0, 1, -1))
```

## Key TakeawaysThe perceptron converges for linearly separable data but cannot solve non-linearly separable tasks such as XOR.
