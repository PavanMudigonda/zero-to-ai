# Bayes' Rule (Basics)

Bayes' rule combines a prior belief with the likelihood of new evidence to obtain a posterior belief.

## Core Idea

For hypothesis $H$ and evidence $E$, $P(H|E)=P(E|H)P(H)/P(E)$. The denominator normalizes the competing hypotheses.

```python
prevalence, sensitivity, false_positive = 0.01, 0.95, 0.05
evidence = sensitivity * prevalence + false_positive * (1 - prevalence)
posterior = sensitivity * prevalence / evidence
print(f'P(condition | positive) = {posterior:.3%}')
```

## Key Takeaways

Likelihood and posterior answer different questions. Base rates matter, even when a test is accurate.
