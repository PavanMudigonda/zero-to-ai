## Core IdeaReplay memory reduces temporal correlation; a target network stabilizes bootstrap targets. Atari preprocessing stacks grayscale frames and often clips rewards.

```python
import random
import torch
from torch import nn
replay = [(torch.randn(4), random.randrange(2), random.random(), torch.randn(4), False) for _ in range(64)]
network = nn.Sequential(nn.Linear(4, 32), nn.ReLU(), nn.Linear(32, 2))
target = nn.Sequential(nn.Linear(4, 32), nn.ReLU(), nn.Linear(32, 2)); target.load_state_dict(network.state_dict())
states = torch.stack([t[0] for t in replay]); actions = torch.tensor([t[1] for t in replay])
rewards = torch.tensor([t[2] for t in replay]); next_states = torch.stack([t[3] for t in replay])
q = network(states).gather(1, actions[:, None]).squeeze(1)
with torch.no_grad(): y = rewards + 0.99 * target(next_states).max(1).values
loss = nn.functional.smooth_l1_loss(q, y); loss.backward()
print('one DQN batch loss:', round(loss.item(), 4))
```

## Key TakeawaysThe small vector example isolates the DQN update. Atari additionally requires an emulator, image preprocessing, and long training.
