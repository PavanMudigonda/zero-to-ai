## Core IdeaAlternating S-cells and C-cells built increasingly complex features with tolerance to small shifts. LeCun's later convolutional networks combined related architectural ideas with supervised backpropagation.

```python
import numpy as np
image = np.zeros((7, 7)); image[1:6, 3] = 1
vertical_filter = np.array([[-1,0,1], [-1,0,1], [-1,0,1]])
response = np.array([[np.sum(image[i:i+3, j:j+3] * vertical_filter) for j in range(5)] for i in range(5)])
pooled = np.array([[response[i:i+2, j:j+2].max() for j in range(0, 4, 2)] for i in range(0, 4, 2)])
print('feature map:\n', response)
print('pooled response:\n', pooled)
```

## Key TakeawaysThe historical connection is conceptual, not identity: modern CNNs differ in training objectives, nonlinearities, and implementation.
