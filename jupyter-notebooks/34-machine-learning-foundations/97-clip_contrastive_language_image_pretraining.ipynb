## Core IdeaA batch forms an image-text similarity matrix. Symmetric cross-entropy treats matching diagonal pairs as targets. Class names can then act as zero-shot text prototypes.

```python
import torch
from torch.nn import functional as F
torch.manual_seed(5); batch, width = 6, 8
image_embeddings = F.normalize(torch.randn(batch, width), dim=1)
text_embeddings = F.normalize(image_embeddings + 0.15 * torch.randn(batch, width), dim=1)
logits = image_embeddings @ text_embeddings.T / 0.07
targets = torch.arange(batch)
loss = (F.cross_entropy(logits, targets) + F.cross_entropy(logits.T, targets)) / 2
print('contrastive loss:', round(loss.item(), 4))
print('image-to-text matches:', logits.argmax(dim=1).tolist())
```
