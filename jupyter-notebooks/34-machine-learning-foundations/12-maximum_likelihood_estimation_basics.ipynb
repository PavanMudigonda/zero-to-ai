## Core IdeaFor Bernoulli observations, the log-likelihood is $\ell(p)=\sum_i[x_i\log p+(1-x_i)\log(1-p)]$, whose maximizer is the sample mean.

```python
import numpy as np

x = np.array([1, 0, 1, 1, 0, 1, 1, 1])
grid = np.linspace(0.01, 0.99, 999)
log_likelihood = np.array([(x * np.log(p) + (1 - x) * np.log(1 - p)).sum() for p in grid])
p_hat = grid[log_likelihood.argmax()]
print(f'Grid MLE: {p_hat:.3f}; analytical MLE: {x.mean():.3f}')
```

## Key TakeawaysLog-likelihood turns products into sums. Many familiar losses are negative log-likelihoods.
