## Core IdeaIf a representative Jacobian factor has magnitude $a$, a path of length $T$ scales approximately as $a^T$. Saturating activations often make $|a|<1$.

```python
import numpy as np
lengths = np.arange(1, 51)
for factor in (0.8, 1.0, 1.2):
    gradients = factor ** lengths
    print(f'factor={factor}: gradient at step 50 = {gradients[-1]:.3e}')
# Clipping limits an exploding gradient's norm without changing its direction.
gradient = np.array([30.0, 40.0])
max_norm = 5.0
clipped = gradient * min(1.0, max_norm / np.linalg.norm(gradient))
print('clipped norm:', np.linalg.norm(clipped))
```

## Key TakeawaysInitialization, normalization, residual paths, clipping, and gated recurrent units improve gradient flow.
