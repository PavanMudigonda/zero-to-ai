## Core IdeaAlexNet combined convolutions, ReLU activations, overlapping max pooling, augmentation, dropout, and multi-GPU training. Full ImageNet training is intentionally replaced here by an architecture inspection.

```python
import torch
from torch import nn
model = nn.Sequential(
    nn.Conv2d(3, 32, 5, stride=2, padding=2), nn.ReLU(), nn.MaxPool2d(3, 2),
    nn.Conv2d(32, 64, 3, padding=1), nn.ReLU(), nn.MaxPool2d(3, 2),
    nn.AdaptiveAvgPool2d((4, 4)), nn.Flatten(), nn.Dropout(0.5), nn.Linear(64 * 4 * 4, 10),
)
x = torch.randn(8, 3, 64, 64)
print('output shape:', tuple(model(x).shape))
print('parameters:', sum(p.numel() for p in model.parameters()))
```
