## Core IdeaNo straight line can separate the two positive XOR corners from the two negative corners. Minsky and Papert analyzed such limits, but the slogan that they proved all neural networks useless is historically misleading.

```python
import numpy as np
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([0, 1, 1, 0])
solutions = []
for w1 in range(-3, 4):
    for w2 in range(-3, 4):
        for b in range(-3, 4):
            pred = (X @ np.array([w1, w2]) + b > 0).astype(int)
            if np.array_equal(pred, y): solutions.append((w1, w2, b))
hidden = np.c_[(X.sum(axis=1) > 0), (X.sum(axis=1) > 1)].astype(int)
multilayer_xor = (hidden[:, 0] - hidden[:, 1] > 0).astype(int)
print('single-layer solutions:', solutions)
print('two-layer XOR:', multilayer_xor)
```

## Key TakeawaysA hidden layer creates intermediate features that make XOR linearly separable in a transformed space.
