## Core IdeaThe update is $c_t=f_t\odot c_{t-1}+i_t\odot\tilde{c}_t$ and $h_t=o_t\odot\tanh(c_t)$. The additive cell-state path helps preserve gradients.

```python
import numpy as np
def sigmoid(x): return 1 / (1 + np.exp(-x))
def lstm_step(x, h, c, W, b):
    gates = np.r_[x, h] @ W + b
    i, f, candidate, o = np.split(gates, 4)
    i, f, o = sigmoid(i), sigmoid(f), sigmoid(o)
    candidate = np.tanh(candidate)
    c = f * c + i * candidate
    return o * np.tanh(c), c
rng = np.random.default_rng(2); input_size = hidden_size = 3
W = rng.normal(scale=0.3, size=(input_size + hidden_size, 4 * hidden_size)); b = np.zeros(4 * hidden_size)
h = c = np.zeros(hidden_size)
for x in np.eye(3): h, c = lstm_step(x, h, c, W, b)
print('hidden:', h.round(3), 'cell:', c.round(3))
```

## Key TakeawaysGates are learned, differentiable controls rather than hard switches. LSTMs mitigate but do not eliminate long-range optimization problems.
