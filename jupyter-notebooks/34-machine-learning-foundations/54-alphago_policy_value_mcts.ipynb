## Core IdeaMCTS balances exploitation and exploration. Policy priors guide promising actions, while value estimates reduce reliance on complete random rollouts.

```python
import math
visits = [12, 3, 0]; value_sum = [8.0, 1.2, 0.0]; priors = [0.55, 0.30, 0.15]
parent_visits, exploration = sum(visits), 1.5
def puct(action):
    mean_value = value_sum[action] / visits[action] if visits[action] else 0.0
    bonus = exploration * priors[action] * math.sqrt(parent_visits) / (1 + visits[action])
    return mean_value + bonus
scores = [puct(action) for action in range(3)]
print('PUCT scores:', [round(score, 3) for score in scores])
print('selected action:', max(range(3), key=lambda action: scores[action]))
```

## Key TakeawaysAlphaGo used supervised expert data before self-play; AlphaGo Zero later learned from self-play without human game records.
