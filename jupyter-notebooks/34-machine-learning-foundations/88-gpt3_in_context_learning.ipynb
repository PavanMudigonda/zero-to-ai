## Core IdeaGPT-1 established generative pretraining plus fine-tuning, GPT-2 emphasized zero-shot transfer, and GPT-3 scaled decoder-only pretraining and demonstrated zero-, one-, and few-shot prompting.

```python
instruction = 'Classify each review as positive or negative.'
examples = [('The film was moving and beautifully acted.', 'positive'), ('The plot was dull.', 'negative')]
query = 'A clever, energetic comedy.'
zero_shot = f'{instruction}\nReview: {query}\nLabel:'
few_shot = instruction + '\n' + '\n'.join(f'Review: {text}\nLabel: {label}' for text, label in examples) + f'\nReview: {query}\nLabel:'
print('ZERO SHOT\n', zero_shot)
print('\nFEW SHOT\n', few_shot)
```

## Key TakeawaysIn-context learning changes behavior through context, not gradient updates. Results depend strongly on wording, examples, order, and model scale.
