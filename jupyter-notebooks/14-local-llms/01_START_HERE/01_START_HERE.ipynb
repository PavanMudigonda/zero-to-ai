# Local LLMs - Start Here

---

This phase is about running language models on your own machine or infrastructure instead of depending entirely on hosted APIs. That matters for privacy, cost control, latency tuning, and deeper understanding of inference tradeoffs.

---

## Why This Phase Matters

Local models are useful when you need more control over deployment, want to experiment without per-token charges, or need workflows that keep data on-device or inside your own environment. They are also one of the best ways to understand practical model serving constraints.

## How To Use This Phase Well

- Treat this phase as systems engineering, not just model shopping.
- Start with easy tooling like Ollama, then compare lower-level or higher-throughput serving options.
- Pay attention to hardware constraints, quantization, latency, and memory tradeoffs.
- Use this phase after you already understand RAG and prompt-based workflows, so you can compare local and hosted setups clearly.
