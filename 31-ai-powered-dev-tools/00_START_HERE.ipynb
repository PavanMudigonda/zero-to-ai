# AI-Powered Dev Tools — Start Here

Welcome to Phase 31 of the Zero-to-AI curriculum. This module teaches you how to turn VS Code into a strongly AI-assisted development environment using GitHub Copilot, MCP (Model Context Protocol), and custom instructions.

**Duration:** 10-15 hours

This phase is optional but useful much earlier than Phase 31 for many learners. If AI-assisted development helps you move faster, you can study it in parallel once you are comfortable with Python, git, and the basic repo structure.

---

## Prerequisites

- Comfortable with Python and VS Code
- Familiarity with git basics
- Enough repo familiarity to understand how AI tooling fits into real workflows

## Key Takeaways

By the end of this module you will be able to:

1. **Explain the 3 layers** of AI-assisted development: completions -> chat -> agent mode
2. **Configure MCP servers** to connect Copilot to databases, browsers, and APIs
3. **Write custom instructions** (`.github/copilot-instructions.md`, `.instructions.md`, `.prompt.md`)
4. **Select and route between models** for different tasks
5. **Build your own MCP server** in Python to expose domain-specific tools
6. **Apply real workflows** for multi-file editing, debugging, testing, and code review
