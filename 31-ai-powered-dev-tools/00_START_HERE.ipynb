# AI-Powered Dev Tools — Start Here

Welcome to Phase 31 of the Zero-to-AI curriculum. This module teaches you to turn **VS Code into a fully AI-native development environment** using GitHub Copilot's agent mode, MCP (Model Context Protocol), and custom instructions.

**Duration:** 10–15 hours (1 week)

---

## Prerequisites

- Comfortable with Python and VS Code
- Familiarity with git basics
- Completed at least Phases 0–8 of the curriculum

## Key Takeaways

By the end of this module you will be able to:

1. **Explain the 3 layers** of AI-assisted development: completions → chat → agent mode
2. **Configure MCP servers** to connect Copilot to databases, browsers, and APIs
3. **Write custom instructions** (`.github/copilot-instructions.md`, `.instructions.md`, `.prompt.md`)
4. **Select and route between models** (GPT-4o, Claude, o3, Gemini) for different tasks
5. **Build your own MCP server** in Python to expose domain-specific tools
6. **Apply real workflows** for multi-file editing, debugging, testing, and code review
