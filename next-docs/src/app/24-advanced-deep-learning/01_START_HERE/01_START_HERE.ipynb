# Advanced Deep Learning - Start Here

---

This phase is an advanced elective for learners who want to go deeper into research-oriented architectures, generative modeling, representation learning, and less standard deep learning topics beyond the core curriculum.

---

## Why This Phase Matters

The earlier neural network phase gives you the core mechanics. This phase expands outward into the deeper model families and research ideas that shaped modern AI systems: GANs, diffusion, advanced transformers, meta-learning, robustness, interpretability, and related topics.

## How To Use This Phase Well

- Do not treat this as a required linear phase for finishing the course.
- Pick a track based on your goals instead of trying to complete all notebooks in order.
- Use it to deepen an area you already understand at the fundamentals level.
- Expect more math, more architectural detail, and less beginner pacing than the core phases.

## Major Tracks

### Generative Models

Focus on GANs, diffusion models, and related generative techniques that underpin modern image and media generation.

### Variational and Probabilistic Methods

Study VAEs, normalizing flows, and uncertainty-aware modeling for a stronger probabilistic view of deep learning.

### Transformer Architectures

Go deeper on BERT, GPT, efficient attention, vision transformers, and scaling patterns such as mixture-of-experts.

### Meta, Continual, and Specialized Learning

Explore few-shot learning, continual learning, curriculum learning, and more research-heavy directions.

## Prerequisites

- Neural Networks (Phase 06)
- Stronger comfort with calculus and linear algebra (Phase 03)
- Working PyTorch knowledge
- Willingness to study selectively rather than linearly

## Suggested Entry Points

- **For generative AI**: `11_diffusion_models.ipynb` then `01_gan_mathematics.ipynb`
- **For LLM internals**: `12_bert_architecture.ipynb` then `13_gpt_architecture.ipynb`
- **For probabilistic ML**: `03_variational_autoencoders_advanced.ipynb` then `37_gaussian_processes.ipynb`

## What Comes Next

Use this phase as a depth amplifier alongside multimodal, fine-tuning, local LLMs, inference optimization, or research-oriented portfolio work. It is best approached as an advanced branch, not as a blocking prerequisite for the rest of the course.
