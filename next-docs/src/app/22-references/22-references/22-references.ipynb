## How to use references well

- Come here when you want a second explanation, more hands-on practice, or a different teaching style for the topic you are already studying.
- Pick only the resource that matches your current phase instead of trying to consume every external collection in parallel.
- After finishing a supplemental resource, return to the repo and apply the extra context in notebooks, quizzes, or project work.
