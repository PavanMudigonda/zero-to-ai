# AI Agents

Learn function calling, ReAct, MCP, agent frameworks, evaluation, and agentic coding workflows as one connected production track.

## Recommended flow

- Start with [Start Here](../01_START_HERE/) and [Intro to Agents](../02_intro_to_agents/) before jumping into frameworks.
- Learn tool use fundamentals through [Function Calling](../03_function_calling/) and [ReAct Pattern](../04_react_pattern/).
- Move into orchestration with [Agent Frameworks](../05_agent_frameworks/), [Multi-Agent Systems](../06_multi_agent_systems/), and [MCP](../07_mcp_model_context_protocol/).
- Finish the main theory track with [Agents SDK and LangGraph](../08_openai_agents_sdk_langgraph/), [Reasoning Models](../09_reasoning_models/), [Autonomous Agents 2026](../10_autonomous_agents_2026/), and [Agent Evaluation](../11_agent_evaluation/).

## Best entry points

- [Start Here](../01_START_HERE/) - orient yourself to the phase and repo flow.
- [MCP](../07_mcp_model_context_protocol/) - the protocol and interoperability layer that matters across modern agent runtimes.
- [Agent Evaluation](../11_agent_evaluation/) - the production bar for judging whether an agent actually works.
- [Agentic Coding IDEs](../13_agentic_coding_ides/) - connect the phase to real coding-agent tooling.

## Practice and assessment

- [Assignment](../12_assignment/) - build a production-ready agent with tools, memory, and evaluation.
- [Challenges](../13_challenges/) - use the challenge set to pressure-test design and debugging choices.
- [Pre-Quiz](../15_pre-quiz/) - assess baseline knowledge.
- [Post-Quiz](../14_post-quiz/) - validate your mastery after completing the track.

## Good project choices

- SQL agent: natural language to queries, results, and insights.
- Research agent: search, synthesize, and report with sources.
- Coding agent: requirements to code, tests, and repair loops.
- Support agent: retrieve context, respond, and escalate when needed.

- [Model Evaluation](../../16-model-evaluation/) - go deeper on metrics, judges, and testing patterns.
- [Debugging and Troubleshooting](../../17-debugging-troubleshooting/) - learn to inspect failures systematically.
- [AI Safety and Red Teaming](../../19-ai-safety-redteaming/) - harden agent systems against misuse and failure.
- [AI-Powered Dev Tools](../../31-ai-powered-dev-tools/) - extend this phase into coding-agent and MCP-heavy workflows.
