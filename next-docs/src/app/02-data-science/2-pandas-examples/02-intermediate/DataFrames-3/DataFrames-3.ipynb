### DataFrame Operations -- Indexing, Renaming, Querying, and Applying Functions

This notebook explores intermediate DataFrame operations using a James Bond films dataset. Topics include: setting and using custom indices with `set_index()`, label-based (`loc[]`) and position-based (`iloc[]`) row access, the deprecated `ix[]` accessor, modifying values in place, renaming columns with `.rename()`, dropping rows and columns with `.drop()`, random sampling with `.sample()`, finding extremes with `.nlargest()`, filtering with `.where()` and `.query()`, applying custom functions with `.apply()`, and understanding the difference between views and copies with `.copy()`. These are the building blocks for constructing complex data transformation pipelines.
