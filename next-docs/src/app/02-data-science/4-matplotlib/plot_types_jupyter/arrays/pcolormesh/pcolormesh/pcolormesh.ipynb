# Pseudocolor Mesh Plots

The `pcolormesh()` method creates a pseudocolor plot on a non-regular rectangular grid, coloring each cell according to the corresponding Z value. Unlike `imshow()`, which assumes evenly spaced pixels, `pcolormesh()` accepts arbitrary X and Y coordinate arrays, allowing it to handle unevenly spaced or even warped grids.

**Why this matters for data science:** Many real-world datasets have non-uniform sampling -- for example, time-series data with irregular timestamps, geographic data in non-Cartesian projections, or experimental data collected at varying resolutions. `pcolormesh()` handles all of these naturally. It is also significantly faster than `pcolor()` for large datasets because it uses flat shading by default. The example demonstrates this with an intentionally uneven x-spacing (note the irregular list of x-values), while y remains uniformly spaced. The `vmin` and `vmax` parameters clamp the color range, which is useful for focusing on a specific value range or ensuring consistent color scales across multiple plots. When comparing `pcolormesh()` vs `imshow()`, choose `pcolormesh()` when your grid coordinates matter or are non-uniform, and `imshow()` when you have a simple matrix to display.

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('_mpl-gallery-nogrid')

# make data with uneven sampling in x
x = [-3, -2, -1.6, -1.2, -.8, -.5, -.2, .1, .3, .5, .8, 1.1, 1.5, 1.9, 2.3, 3]
X, Y = np.meshgrid(x, np.linspace(-3, 3, 128))
Z = (1 - X/2 + X**5 + Y**3) * np.exp(-X**2 - Y**2)

# plot
fig, ax = plt.subplots()

ax.pcolormesh(X, Y, Z, vmin=-0.5, vmax=1.0)

plt.show()
```
