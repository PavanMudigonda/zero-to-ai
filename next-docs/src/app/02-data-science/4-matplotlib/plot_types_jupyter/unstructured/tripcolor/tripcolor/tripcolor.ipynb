# Pseudocolor Plots on Unstructured Grids

The `tripcolor()` method creates a pseudocolor (false-color) plot on an unstructured triangular grid, where each triangle in the Delaunay triangulation is colored according to the z-values at its vertices. Unlike `tricontourf()` which interpolates smooth contour boundaries, `tripcolor()` colors each triangle individually, making the triangulation itself visible in the output.

**Why this matters for data science:** `tripcolor()` is the unstructured-data equivalent of `pcolormesh()` and is particularly useful when you want to see the resolution of your triangulation alongside the data values. In finite element analysis, computational fluid dynamics, and structural engineering simulations, the mesh itself carries important information about where the computation focused its resolution. The example shows the raw data points as grey dots alongside the colored triangulation, making it clear how the Delaunay algorithm connects scattered points. The visible triangle edges in the output distinguish `tripcolor()` from the smoother `tricontourf()` -- choose `tripcolor()` when mesh fidelity matters, and `tricontourf()` when you want a polished continuous appearance.

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('_mpl-gallery-nogrid')

# make data:
np.random.seed(1)
x = np.random.uniform(-3, 3, 256)
y = np.random.uniform(-3, 3, 256)
z = (1 - x/2 + x**5 + y**3) * np.exp(-x**2 - y**2)

# plot:
fig, ax = plt.subplots()

ax.plot(x, y, 'o', markersize=2, color='grey')
ax.tripcolor(x, y, z)

ax.set(xlim=(-3, 3), ylim=(-3, 3))

plt.show()
```
