# Fill Between Curves

The `fill_between()` method shades the region between two curves (y1 and y2), creating a band that visually emphasizes the gap or range between them. Combined with a line plot of the midpoint, it produces the classic "confidence band" visualization.

**Why this matters for data science:** Fill-between plots are ubiquitous in data science for showing uncertainty, variability, and confidence intervals. You will see them in time-series forecasting (prediction intervals around a forecast), A/B test results (confidence bands around conversion rates), and model evaluation (showing variance across cross-validation folds). The `alpha=0.5` parameter makes the fill semi-transparent, which is critical when overlaying bands on top of line plots so both remain visible. The example generates two noisy upward-trending curves using `np.random.uniform()` for the noise component, then plots the midpoint `(y1 + y2)/2` as a solid line through the center of the band. Setting `linewidth=0` on the fill removes the border lines of the filled region for a cleaner look.

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('_mpl-gallery')

# make data
np.random.seed(1)
x = np.linspace(0, 8, 16)
y1 = 3 + 4*x/8 + np.random.uniform(0.0, 0.5, len(x))
y2 = 1 + 2*x/8 + np.random.uniform(0.0, 0.5, len(x))

# plot
fig, ax = plt.subplots()

ax.fill_between(x, y1, y2, alpha=.5, linewidth=0)
ax.plot(x, (y1 + y2)/2, linewidth=2)

ax.set(xlim=(0, 8), xticks=np.arange(1, 8),
       ylim=(0, 8), yticks=np.arange(1, 8))

plt.show()
```
