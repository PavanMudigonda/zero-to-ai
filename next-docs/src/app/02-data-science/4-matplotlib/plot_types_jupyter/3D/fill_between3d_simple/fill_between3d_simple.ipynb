# 3D Fill Between

The 3D `fill_between()` method shades the region between two 3D curves, creating a surface that connects them. Unlike its 2D counterpart which fills between a curve and a baseline, the 3D version takes two complete sets of coordinates (x1, y1, z1) and (x2, y2, z2), and fills the space between those two paths in three-dimensional space.

**Why this matters for data science:** Filled regions between 3D curves are valuable for visualizing confidence intervals in three-dimensional trajectory data, tolerance bands around predicted flight paths, or the spread between two related time-evolving signals. The example here constructs a double helix using parametric equations with `np.cos()` and `np.sin()`, offset by pi radians -- a pattern that appears frequently when visualizing phase relationships in signal processing or molecular structures. The `alpha=0.5` parameter controls transparency, which is essential in 3D plots to prevent the filled surface from completely hiding the underlying line plots.

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('_mpl-gallery')

# Make data for a double helix
n = 50
theta = np.linspace(0, 2*np.pi, n)
x1 = np.cos(theta)
y1 = np.sin(theta)
z1 = np.linspace(0, 1, n)
x2 = np.cos(theta + np.pi)
y2 = np.sin(theta + np.pi)
z2 = z1

# Plot
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.fill_between(x1, y1, z1, x2, y2, z2, alpha=0.5)
ax.plot(x1, y1, z1, linewidth=2, color='C0')
ax.plot(x2, y2, z2, linewidth=2, color='C0')

ax.set(xticklabels=[],
       yticklabels=[],
       zticklabels=[])

plt.show()
```
