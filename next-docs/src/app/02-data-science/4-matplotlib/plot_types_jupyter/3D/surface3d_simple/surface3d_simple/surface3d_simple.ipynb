# 3D Surface Plots

A surface plot renders a continuous 2D surface in 3D space, where the height (z-value) at each point is determined by a function of x and y. The `Axes3D.plot_surface()` method takes 2D arrays X, Y, and Z (typically generated with `np.meshgrid()`) and renders the surface with optional colormapping to enhance depth perception.

**Why this matters for data science:** Surface plots are the go-to visualization for understanding loss landscapes in machine learning -- plotting how a cost function varies across two parameters helps build intuition about optimization challenges like local minima, saddle points, and flat regions. They are also used for visualizing probability density functions, response surfaces in experimental design, and interpolated terrain data. The example computes Z as the sine of the radial distance from the origin (`np.sqrt(X**2 + Y**2)`), producing a classic "ripple" pattern. The `cmap=cm.Blues` parameter applies a colormap that reinforces the height information visually, and `vmin=Z.min() * 2` adjusts the color scaling to emphasize the surface contours.

```python
import matplotlib.pyplot as plt
import numpy as np

from matplotlib import cm

plt.style.use('_mpl-gallery')

# Make data
X = np.arange(-5, 5, 0.25)
Y = np.arange(-5, 5, 0.25)
X, Y = np.meshgrid(X, Y)
R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)

# Plot the surface
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.plot_surface(X, Y, Z, vmin=Z.min() * 2, cmap=cm.Blues)

ax.set(xticklabels=[],
       yticklabels=[],
       zticklabels=[])

plt.show()
```
