# Practical Data Science

This phase is where you stop sampling topics and start doing end-to-end work. Treat it like applied training for interviews, take-homes, portfolio projects, and real delivery work rather than as another theory-heavy chapter.

## How To Use This Phase Well

- Pick a role-aligned subtrack instead of trying to complete every folder.
- Optimize for one or two strong projects with written tradeoffs, not checkbox coverage.
- Treat communication, evaluation, and reproducibility as part of the project, not optional extras.
- Revisit earlier phases as needed when a project exposes a weakness in your foundations.

## Recommended Order

1. Read `README.md` in this folder to choose the most relevant subtrack.
2. Complete `00_INTERVIEW_PREP.ipynb` to set expectations for practical work.
3. Work through one role-aligned subtrack deeply.
4. Finish with one capstone-quality project that includes a metric, tradeoffs, and a short written explanation.

## Good Study Routes

- **ML Engineer**: `machine-learning/`, `statistics-mlops/`, `sql-data-engineering/`
- **Data Scientist**: `python-data-science/`, `time-series-forecasting/`, `recommender-causal/`
- **Applied AI Engineer**: `deep-learning-nlp/`, `computer-vision/`, `machine-learning/`

The goal is depth with evidence, not checkbox completion.

## What Comes Next

After this phase, turn your strongest project into a portfolio piece, case study, or interview story. This is one of the best places in the repo to produce proof of work rather than just study notes.
